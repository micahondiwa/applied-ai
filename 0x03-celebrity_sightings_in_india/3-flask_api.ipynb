# **Making a Web App**

**Project Goal**: To create a web app to run the face recognition facility. This will create a tool that anyone can upload a file to. The app will give them back an image with faces outlined and labeled, at least for faces that it knows about.

**Objectives**

- Make a series of functions we can call to do face recognition.
- Create a `py` file so our app can use those functions.
- Assemble a `Flask` app as a front-end to our face recognizer.
- Test the app.
